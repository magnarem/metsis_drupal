```python
import json

# Assuming your JSON file is named 'data.json' and is in the same directory
file_name = 'simple_metsis_documents.json' 

try:
    with open(file_name, 'r') as f:
        data = json.load(f)
    pretty_json_string = json.dumps(data, indent=2) # Using 2 spaces for indentation
    print(pretty_json_string)
except FileNotFoundError:
    print(f"Error: The file '{file_name}' was not found in the current directory.")
except json.JSONDecodeError:
    print(f"Error: Could not decode JSON from '{file_name}'. Check file format.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```
